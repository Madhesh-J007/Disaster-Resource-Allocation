# Notebook 01: Exploratory Dataset Analysis (EDA)
## Secondary Disaster Chain Reaction ML System

This notebook inspects all raw integrated datasets (EM-DAT, NASA EONET, ReliefWeb, OpenStreetMap, IMD Rainfall, WorldPop, Census, SRTM DEM).


```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = r'c:\Project\ML\Disaster-Resource-Allocation-ML'
RAW_DIR = os.path.join(BASE_DIR, 'datasets', 'raw')

print('Base directory configured:', BASE_DIR)
```

### 1. Inspect EM-DAT Disaster History Records


```python
df_emdat = pd.read_csv(os.path.join(RAW_DIR, 'emdat', 'emdat_disasters.csv'))
print(f'EM-DAT Shape: {df_emdat.shape}')
display(df_emdat[['Year', 'Disaster Type', 'Location', 'Total Deaths', 'Total Affected', 'Total Damages ('000 US$)']].head())
```

### 2. Inspect OpenStreetMap Tamil Nadu Infrastructure Nodes


```python
with open(os.path.join(RAW_DIR, 'osm', 'tamil_nadu_infrastructure.json'), 'r', encoding='utf-8') as f:
    osm_data = json.load(f)

elements = osm_data.get('elements', [])
print(f'Total OpenStreetMap Elements Loaded: {len(elements)}')

tags_list = [el.get('tags', {}).get('amenity') or el.get('tags', {}).get('power') for el in elements if el.get('type') == 'node']
df_osm_tags = pd.Series(tags_list).value_counts().reset_index()
df_osm_tags.columns = ['Infrastructure Type', 'Count']
display(df_osm_tags)
```

### 3. Inspect IMD Multi-Year Weather & Precipitation Observations


```python
df_imd = pd.read_csv(os.path.join(RAW_DIR, 'imd', 'india_rainfall.csv'))
print(f'IMD Rainfall Observations Shape: {df_imd.shape}')
display(df_imd.describe())
```

### 4. Inspect District Demographics & Topography (WorldPop, Census, DEM)


```python
df_pop = pd.read_csv(os.path.join(RAW_DIR, 'worldpop', 'worldpop_tamil_nadu.csv'))
df_census = pd.read_csv(os.path.join(RAW_DIR, 'census', 'census_tamil_nadu.csv'))
df_dem = pd.read_csv(os.path.join(RAW_DIR, 'dem', 'tamil_nadu_dem.csv'))

df_merged_eda = pd.merge(df_pop, df_census, on='district')
df_merged_eda = pd.merge(df_merged_eda, df_dem[['district', 'elevation_m']], on='district')
print(f'Merged District EDA Table Shape: {df_merged_eda.shape}')
display(df_merged_eda[['district', 'population_2020', 'density_per_sq_km', 'urban_pct', 'slum_pop_pct', 'elevation_m']].head(10))
```
