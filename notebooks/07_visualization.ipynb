# Notebook 07: Spatial GIS & Interactive Dashboard Visualizations

Plots spatial district risk maps, precipitation heatmaps, and interactive dashboard components.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

BASE_DIR = r'c:\Project\ML\Disaster-Resource-Allocation-ML'
f_path = os.path.join(BASE_DIR, 'datasets', 'processed', 'feature_engineered_dataset.csv')
df = pd.read_csv(f_path)

plt.figure(figsize=(12, 6))
sns.barplot(data=df.sort_values('cascade_risk_score', ascending=False), x='district', y='cascade_risk_score', palette='magma')
plt.title('Secondary Disaster Cascade Risk Score by Tamil Nadu District', fontsize=14, fontweight='bold')
plt.xticks(rotation=90, fontsize=8)
plt.ylabel('Cascade Risk Score (0-100)')
plt.tight_layout()
plt.show()
```
