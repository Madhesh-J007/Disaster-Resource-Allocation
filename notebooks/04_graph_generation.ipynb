# Notebook 04: Infrastructure Dependency Graph & Cascade Simulation

Constructs NetworkX graph nodes (hospitals, power substations, fire stations, road bridges) and simulates secondary failure cascades.


```python
import os
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# Create Infrastructure Dependency Network Graph
G = nx.DiGraph()

# Add Nodes
nodes = [
    ('Chennai_Substation_01', {'type': 'Power', 'capacity': 100}),
    ('Rajiv_Gandhi_Hospital', {'type': 'Hospital', 'backup_gen': True}),
    ('Chennai_Central_Fire', {'type': 'Emergency', 'units': 15}),
    ('Kaveri_Bridge_Node', {'type': 'Transport', 'status': 'Operational'}),
    ('Cuddalore_Substation', {'type': 'Power', 'capacity': 80}),
    ('Cuddalore_Govt_Hospital', {'type': 'Hospital', 'backup_gen': False})
]

for n, attrs in nodes:
    G.add_node(n, **attrs)

# Add Directed Dependencies (Power -> Hospital, Transport -> Emergency)
edges = [
    ('Chennai_Substation_01', 'Rajiv_Gandhi_Hospital', {'weight': 0.9, 'dependency': 'Power'}),
    ('Chennai_Substation_01', 'Chennai_Central_Fire', {'weight': 0.8, 'dependency': 'Power'}),
    ('Kaveri_Bridge_Node', 'Rajiv_Gandhi_Hospital', {'weight': 0.95, 'dependency': 'Access Road'}),
    ('Cuddalore_Substation', 'Cuddalore_Govt_Hospital', {'weight': 0.9, 'dependency': 'Power'}),
]

for u, v, attrs in edges:
    G.add_edge(u, v, **attrs)

print(f'Dependency Graph: {G.number_of_nodes()} Nodes, {G.number_of_edges()} Edges')
```

### Simulate Secondary Failure Cascade


```python
def simulate_cascade(failed_node):
    print(f'Triggering Primary Failure at Node: {failed_node}')
    failed = {failed_node}
    secondary = set()
    for u, v, d in G.edges(data=True):
        if u == failed_node:
            secondary.add(v)
            print(f'  -> Secondary Impact: {v} disrupted via {d["dependency"]} link!')
    return failed, secondary

failed, secondary = simulate_cascade('Chennai_Substation_01')
```
