# Notebook 05: Machine Learning Model Training

Trains Random Forest and XGBoost Regressors / Classifiers to predict Cascade Risk Score and Risk Level.


```python
import os
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score, classification_report

BASE_DIR = r'c:\Project\ML\Disaster-Resource-Allocation-ML'
f_path = os.path.join(BASE_DIR, 'datasets', 'processed', 'feature_engineered_dataset.csv')
df = pd.read_csv(f_path)

# Features & Target
ignore_cols = ['district_id', 'district', 'cascade_risk_score', 'cascade_risk_level', 'secondary_disaster_triggered']
num_cols = [c for c in df.columns if c not in ignore_cols and pd.api.types.is_numeric_dtype(df[c])]

X = df[num_cols]
y_reg = df['cascade_risk_score']
y_clf = df['cascade_risk_level']

X_train, X_test, y_train_reg, y_test_reg = train_test_split(X, y_reg, test_size=0.2, random_state=42)
_, _, y_train_clf, y_test_clf = train_test_split(X, y_clf, test_size=0.2, random_state=42)

print(f'Train set: {X_train.shape}, Test set: {X_test.shape}')
```

### Train Random Forest Regressor & Classifier


```python
rf_reg = RandomForestRegressor(n_estimators=100, random_state=42)
rf_reg.fit(X_train, y_train_reg)
y_pred_reg = rf_reg.predict(X_test)
print('Random Forest Regressor R2 Score:', round(r2_score(y_test_reg, y_pred_reg), 4))
print('Random Forest Regressor RMSE:', round(np.sqrt(mean_squared_error(y_test_reg, y_pred_reg)), 4))

rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
rf_clf.fit(X_train, y_train_clf)
y_pred_clf = rf_clf.predict(X_test)
print('Random Forest Classifier Accuracy:', round(accuracy_score(y_test_clf, y_pred_clf), 4))
print('\nClassification Report:\n', classification_report(y_test_clf, y_pred_clf, zero_division=0))
```
